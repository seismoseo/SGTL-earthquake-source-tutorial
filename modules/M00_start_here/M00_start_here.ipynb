## 1 · Which machine runs what

You work on two computers at once. It pays to know which one does what.

- Your **laptop** runs the web browser, plus a window for **SSH** (secure shell: a secure way to log in to
  another computer over the network and run commands there).
- The **server** (a shared computer that many people log in to and run programs on at the same time) runs
  everything else: **JupyterLab** (a program that runs in your web browser and lets you write and run notebooks),
  the **kernel** (the program behind a notebook that runs your code cells; here it is "SGTL (Python 3.12)"), and
  it stores your files.
- The **SSH tunnel** (a forwarded port that lets the browser on your laptop reach a program on the server; a
  *port* is a numbered door through which programs talk over a network) joins the two. It is the laptop window
  that seems to hang while you work: it must stay open.

| What you see | Where it actually runs |
|---|---|
| the browser tab with JupyterLab | the server (your laptop only draws the page) |
| the output under a code cell | the server (the kernel) |
| a terminal opened with File → New → Terminal in JupyterLab | the server |
| the window where you started the tunnel | your laptop |
| the files in JupyterLab's file browser (left) | the server, in your folders there |

Two consequences. Closing the browser tab does not stop the kernel: it keeps running, and keeps its memory, on the
server until you shut it down (Kernel → Shut Down Kernel). And a file on your laptop is not on the server: copy it
over before you use it.

The course software lives in an **environment** (a self-contained set of Python and packages with fixed versions,
so results do not change when other software does) that **conda** (a package manager that installs Python and
scientific libraries into separate environments) built under `~/sgtl/`. The kernel uses that environment.
Installation and the tunnel recipes are in [docs/setup.md](../../docs/setup.md).

### Just enough Python for M00

Your answers in this module are Python **dicts**. A dict maps *keys* to *values*: `{"key": value, ...}`. Text (a
*string*) goes in quotes, numbers do not, a list goes in square brackets, and `None` means "no value". A
**function** (a named block of code that takes inputs and returns an output) starts with `def`; whatever follows
`return` is its result. In every task you replace two lines, `# YOUR CODE HERE` and `raise NotImplementedError(...)`,
with your own lines and keep everything else. A worked example with made-up values:

````python
def demo_station():
    """A made-up station, to show the parts of a dict (nothing here is real)."""
    station = {"code": "DEMO1", "elevation_m": 120, "latitude": 35.1, "channels": ["HHZ", "HHN", "HHE"],
               "closed_on": None}
    return station


info = demo_station()
print(info["code"], info["elevation_m"])          # read one value by its key
print(type(info["code"]), type(info["elevation_m"]), type(info["latitude"]), type(info["channels"]))
````

Three more pieces you will meet often:

- **Powers.** `x ** 2` is x squared (x to the power 2), and `x ** 0.5` is its square root. Careful: `^` is
  **not** a power in Python. `3 ^ 2` gives 1, not 9 (it combines the binary digits of two whole numbers), and
  Python shows no error, so the slip is easy to miss.
- **Libraries.** `import numpy as np` loads a **library** (a package of ready-made functions that someone else
  wrote; NumPy is the one for arrays of numbers) under the short name `np`, so `np.sqrt(2)` means "the function
  `sqrt` of NumPy". The setup cell's `import sgtl_tutorial as st` works the same way.
- **Methods.** A **method** is a function that belongs to a value. You call it with a dot after the value and
  brackets at the end: `obj.method()`. `"DEMO1".lower()` gives `"demo1"`, and `info.keys()` gives the keys of
  the dict `info`. Without the brackets you get the method itself, not its result.

````python
a = 3
print(a ** 2, a ^ 2)              # 9, then 1: only ** is a power
import numpy as np                # the library NumPy, under its short name np
print(np.sqrt(a ** 2))            # a function of the library: short name, dot, function name, brackets
print(info["code"].lower())       # a method of the string "DEMO1"
print(list(info.keys()))          # a method of the dict info (list(...) shows its result as a plain list)
````

## 4 · Looking inside files from the shell

Data in this course often comes as **CSV** files (comma-separated values: a plain-text table with one row per
line and the columns separated by commas; the first line, the *header*, usually holds the column names). Many
questions about such a file can be answered without Python, with small commands joined by a **pipe** `|`, which
feeds what one command prints into the next command, for example `ls -l | head -n 3`.

| Command | What it does |
|---|---|
| `head -n 3 FILE`, `tail -n 3 FILE` | print the first or the last 3 lines |
| `tail -n +2 FILE` | print everything from line 2 on (skips a header line) |
| `wc -l FILE` | count the lines (`wc -c` counts bytes) |
| `cut -d, -f2 FILE` | keep only column 2, splitting each line at the commas |
| `grep 'text' FILE` | keep the lines that contain `text` anywhere; `grep -c` counts them; the pattern `'^text'` keeps only lines that *start* with `text` |
| `sort`, `sort -n` | sort lines as text, or as numbers |
| `uniq -c` | merge identical neighboring lines and count them (so sort first) |
| `ls -l FILE`, `ls -lh FILE` | the size in bytes, or in readable units (K = 1024 bytes) |
| `du -h FILE` | the disk space the file takes (counted in whole storage blocks, so a little more) |

A worked example on a tiny made-up file (these are not real stations). The next cell writes the file into the
notebook's folder; the cell after it runs some of the commands above. Before you run it, guess each output.

## 5 · Sharing the server

The server is shared, and one careless command can slow down or kill everyone's work. Three words:

- a **CPU core** is one processing unit of a computer; all users of a shared server share a fixed number of them;
- a **thread** is one stream of work inside a running program; more threads use more CPU cores at the same time.
  NumPy does its heavy arithmetic with a **BLAS** library (Basic Linear Algebra Subprograms, the engine behind
  matrix arithmetic), which by default starts one thread per core: enough to take over the whole machine;
- **memory** (RAM) is the fast working storage of running programs; when the server runs out, programs are killed,
  possibly someone else's.

The setup cell at the top called `st.init`, which capped this notebook at `CORES` thread(s) per library and set a
*soft memory cap* of `MEM_GB` GB: a limit the kernel puts on itself, so that a runaway cell stops with a
`MemoryError` instead of hurting others. For work in a terminal the rules are in
[docs/server_manners.md](../../docs/server_manners.md); the short version:

```bash
export OMP_NUM_THREADS=1 OPENBLAS_NUM_THREADS=1 MKL_NUM_THREADS=1 NUMEXPR_NUM_THREADS=1 NUMPY_MADVISE_HUGEPAGE=0
nice -n 15 python my_script.py     # nice lowers the priority, so people who are typing go first
nproc; uptime; free -g             # how many cores, how busy the server is, how much memory is free
top -u $USER                       # your own programs (q quits)
```

The next cell prints a dict that describes the live limits of this kernel. Nothing in it is secret, but read it

## 6 · Git: snapshots, and where they go

**git** is a version-control program that records snapshots of your files, so you can compare and undo changes.
A **repository** is a folder whose full history of changes git tracks; your copy of the tutorial is one. A
**commit** is one saved snapshot, with a short message that says what changed and why. A **branch** is a named
line of commits (yours is `main` unless you make another one).

A repository can know other copies of itself, called *remotes*, each under a short name:

- **upstream**: the original repository your copy came from, here the public tutorial
  `github.com/seismoseo/SGTL-earthquake-source-tutorial`. You only fetch tutorial updates from it (your mentor
  announces them; `bin/update-from-upstream` merges them).
- **origin**: your **private copy** (your own private GitHub repository that keeps the tutorial's history and holds
  your work). You *push* (upload) your commits there; only you and your mentor can see them.

The daily cycle, in a terminal in the top folder of your copy:

```bash
git status                               # what changed since the last commit?
git add work/M00_start_here/             # choose what goes into the next snapshot
git commit -m "M00: first two tasks"     # take the snapshot, with a message
git push                                 # upload the new commits to origin
git pull                                 # download new commits from origin (after working somewhere else)
git log --oneline                        # the history, newest first: a short hash, then the message
```

The *hash* is a code of letters and digits that names one commit; its first 7 characters are enough to point at
it. Commit small steps often, with messages that say why: version control from the first day is one of the
"good enough practices" for research code (Wilson et al., 2014, 2017).

Right after cloning, `origin` still points to the public tutorial. `bin/make-private-copy` fixes that, once. Like
every course script it is a **dry run** by default (it only prints what it would do, without changing anything);
`--apply` makes it act. You need a GitHub account with **two-factor authentication** (logging in with your password
plus a one-time code from your phone, so a stolen password alone is not enough) and a GitHub login on the server
(`gh auth login`, [docs/setup.md](../../docs/setup.md) step 7). Then:

```bash
bin/make-private-copy              # dry run: read every line it prints
bin/make-private-copy --apply      # make your private repository and rewire the remotes
git remote -v                      # origin = your private copy, upstream = the public tutorial
```

More in [docs/git_workflow.md](../../docs/git_workflow.md) and the Software Carpentry lesson *Version Control with
Git*.

## 7 · AI as a tutor, and the AI log

You may use AI tools (Claude, ChatGPT, Copilot and others) **as a tutor, not as an answer key**
([docs/ai_policy.md](../../docs/ai_policy.md); Korean: [docs/ai_policy_ko.md](../../docs/ai_policy_ko.md)).

- **Allowed:** asking what an error means; asking for explanations of concepts, commands and library functions;
  asking for a review of code *you* wrote; asking for test ideas; general Python, shell and git questions with
  generic examples.
- **Not allowed:** letting AI write, complete or fix a task function, a prediction or an Explain answer; pasting
  check output into an AI and trying its ideas until the check passes; pasting passwords, tokens, your JupyterLab
  link or data files into a chat.

Why so strict? Studies of students who learned with AI found that unrestricted help can raise scores during
practice but lower what students can do on their own afterwards, and that it can leave beginners with an
*illusion of competence*: the feeling of understanding code they could not have written (Bastani et al., 2025;
Prather et al., 2024; Kazemitabaar et al., 2023). The weekly oral check asks you to explain your own cells, so
write code you can explain.

The **AI log** is one Markdown file per module in `work/ai_log/` (Markdown is plain text in which a line starting
with `#` is a heading and a line starting with `-` is a bullet point); `bin/sgtl new M00` already made
`work/ai_log/M00.md` from the template. Each entry gives the date, task id and tool, then what you asked, what
the AI answered, what you used and changed, whether the help touched a task function, prediction or explanation,
and what you learned. An honest log never counts against you; a missing one fails the gate.
