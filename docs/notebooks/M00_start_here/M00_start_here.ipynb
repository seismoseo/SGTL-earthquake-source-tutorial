## Part 4 · Looking inside a CSV file (about 30 min, then task 2)

A **CSV file** is a text table: one row per line, commas between the columns, and the column names in the first
line, the *header*. A **pipe** `|` feeds what one command prints into the next: `ls -l | head -n 3`.

| Command | What it prints |
|---|---|
| `head -n 3 FILE`, `tail -n 3 FILE` | the first or last 3 lines |
| `tail -n +2 FILE` | the lines from line 2 on |
| `wc -l FILE` | the number of lines (`-c`: of bytes) |
| `cut -d, -f2 FILE` | column 2, split at commas |
| `grep 'text' FILE` | lines that contain `text` (`-c` counts them; `'^text'`: at the start only) |
| `sort`, `sort -n` | lines sorted as text, or as numbers |
| `uniq -c` | identical neighbor lines, counted |
| `ls -l FILE`, `ls -lh FILE` | the size in bytes, or in K (1024 bytes) |

Guess what each command in the next two cells prints, then run them.

````python
demo_csv = """station,network,latitude,longitude,elevation_m
DEMO1,XX,35.10,129.00,120
DEMO2,XX,35.40,128.70,85
DEMO3,YY,36.00,127.50,310
DEMO4,XX,34.90,126.80,45
DEMO5,YY,37.20,127.10,150
"""
with open("demo_stations.csv", "w", encoding="utf-8") as fh:
    fh.write(demo_csv)
print("wrote demo_stations.csv into the notebook's folder")
````

## Part 6 · Git and your private copy (about 15 min, then task 4)

tracks, like your copy; a **commit** is one snapshot. A **remote** is another copy: **upstream**, the public
tutorial, sends you updates; **origin**, your **private copy** on GitHub, keeps the commits you **push**. A
**branch** is a named line of commits; yours is `main`.

course scripts it is a **dry run** by default: it only prints what it would do, and `--apply` acts. It needs your

